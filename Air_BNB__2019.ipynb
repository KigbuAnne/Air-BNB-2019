HANDLING MISSING DATA

 1.Replace missing name and host_name with "Unknown".

```python
df[['name', 'host_name']] = df[['name', 'host_name']].fillna("Unknown")
```

Fill reviews_per_month with 0 if no reviews:

```python
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
```

Fill missing last_review with a placeholder or drop it:


```python
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')
```

CHECK FOR DUPLICATES

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

CONVERTING DATA TYPES

converting categorical varible like neighbourhood_group and room_type.

```python
df['neighbourhood_group'] = df['neighbourhood_group'].astype('category')
df['room_type'] = df['room_type'].astype('category')
```

SAVE THE CLEANED DATASET


```python
df.to_csv("cleaned_airbnb_nyc_2019.csv", index=False)
```
