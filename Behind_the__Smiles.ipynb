```python
# Load the Excel file
file_path = "Life Expectancy Data 2021.xlsx"
df = pd.read_excel(file_path, sheet_name='Life Expectancy 2021')

# Filter for total population and relevant indicator
filtered_df = df[(df['DIM_SEX'] == 'TOTAL') &
                 (df['IND_NAME'] == 'Life expectancy (at birth)')]

# Sort and get top 10 countries
top_10 = filtered_df.sort_values(by='AMOUNT_N', ascending=False).head(10)
top_10 = top_10[['COUNTRY', 'AMOUNT_N']].rename(columns={'COUNTRY': 'Country', 'AMOUNT_N': 'Life Expectancy'})

# Plotting
plt.figure(figsize=(12, 6))
bars = plt.bar(top_10['Country'], top_10['Life Expectancy'], color='skyblue')
plt.xlabel('Country')
plt.ylabel('Life Expectancy (Years)')
plt.title('Top 10 Countries by Life Expectancy in 2021')
plt.xticks(rotation=45)
plt.ylim(80, 85)

# Annotate bars with values
```

From the bar chart, we see that the countries that are high in Life expectancy are also countries that are among the top countries in Happiness. Norway,Luxembourg,Sweden,Iceland,are among the top 10 happiest countries in the world.  

```python
# Load the Excel file
df = pd.read_excel('Life Expectancy Data 2021.xlsx')

# Filter for total population and relevant indicator
filtered_df = df[(df['DIM_SEX'] == 'TOTAL') &
                 (df['IND_NAME'] == 'Life expectancy (at birth)')]

# Get the 10 countries with the lowest life expectancy
bottom_10 = filtered_df.sort_values(by='AMOUNT_N', ascending=True).head(10)
bottom_10 = bottom_10[['COUNTRY', 'AMOUNT_N']].rename(columns={'COUNTRY': 'Country', 'AMOUNT_N': 'Life Expectancy'})

# Set the plot style
sns.set(style="whitegrid")

# Plot using seaborn
plt.figure(figsize=(12, 6))
barplot = sns.barplot(data=bottom_10, x='Country', y='Life Expectancy', palette='Reds_r')

# Add annotations on bars
for p in barplot.patches:
    barplot.annotate(f'{p.get_height():.2f}',
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='bottom', fontsize=9)

# Customize the plot
plt.title('Bottom 10 Countries by Life Expectancy in 2021', fontsize=14)
plt.xlabel('Country')
plt.ylabel('Life Expectancy (Years)')
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

From the bar chart above,Lesotho,Zimbabwe, Afghanistan are among the nations that are least happy 10 countres in the world as well as the least in Life Expectancy.

## THE RELATIONSHIP

Using a scatter plot,lets visualize the relationship between life expectancy and high happiness.

```python
# Load dataset
data = {
    'Country': ['Finland', 'Denmark', 'Iceland', 'Sweden', 'Netherlands'],
    'Happiness Score': [7.736, 7.521, 7.515, 7.345, 7.306],
    'Life Expectancy': [82.66, 82.57, 81.53, 81.18, 81.12]
}
df = pd.DataFrame(data)

# Compute correlation
correlation = df['Happiness Score'].corr(df['Life Expectancy'])
print(f'Correlation between Happiness Score and Life Expectancy: {correlation:.2f}')

# Visualize the relationship
sns.set(style="whitegrid")
plt.figure(figsize=(8, 6))
sns.regplot(x='Life Expectancy', y='Happiness Score', data=df, marker='o')
plt.title('Happiness Score vs Life Expectancy')
plt.xlabel('Life Expectancy (Years)')
plt.ylabel('Happiness Score')
plt.tight_layout()
plt.show()
```

There's a positive correlation between the hapiness score and an increase in life expectancy.

```python
# Load the dataset, making sure the file path is correct
# If the file is in the same directory as the notebook, you can use just the filename
# Otherwise, provide the full path to the file
df = pd.read_excel("world_happiness_report 2024.xlsx", sheet_name="Data for Figure 2.1 (2011–2024)") # Changed file name

# Filter for the year 2024
df_2024 = df[df["Year"] == 2024]

# Select relevant columns
columns = [
    "Ladder score",
    "Explained by: Log GDP per capita",
    "Explained by: Social support",
    "Explained by: Healthy life expectancy",
    "Explained by: Freedom to make life choices",
    "Explained by: Generosity",
    "Explained by: Perceptions of corruption"
]

# Rename columns for clarity
df_2024 = df_2024[columns].rename(columns={
    "Ladder score": "Happiness",
    "Explained by: Log GDP per capita": "GDP per Capita",
    "Explained by: Social support": "Social Support",
    "Explained by: Healthy life expectancy": "Life Expectancy",
    "Explained by: Freedom to make life choices": "Freedom",
    "Explained by: Generosity": "Generosity",
    "Explained by: Perceptions of corruption": "Corruption"
})

# Compute the correlation matrix
corr_matrix = df_2024.corr()

# Plot the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap - World Happiness Report 2024")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

Based on the pairplot and heatmap correlation analysis, several key insights emerge regarding the factors influencing Happiness scores:

GDP per Capita, Social Support, and Life Expectancy exhibit strong positive correlations with Happiness. This suggests that countries with higher economic performance, stronger social networks, and better health outcomes tend to report higher levels of happiness.

Freedom to Make Life Choices shows a moderate positive correlation, indicating it contributes to happiness, though to a lesser extent than the core economic and social factors.

Generosity and Perceptions of Corruption demonstrate weak or no significant correlation with Happiness. These results imply that while these variables may reflect societal values or governance, they are not strong predictors of overall happiness in the dataset.

These findings underscore the importance of economic stability, health, and social support in shaping national well-being.
